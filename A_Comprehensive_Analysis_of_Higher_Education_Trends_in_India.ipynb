## Problem Statement:

This project focuses on analyzing higher education data in India from 2013 to 2021 across Engineering, Pharmacy, and Management institutions. The dataset contains information on the number of institutions, student intake, enrollment, faculty strength, pass-out rates, placements, and institutional growth. Using Data Analytics techniques such as data cleaning, exploratory data analysis (EDA), and data visualization, the project aims to identify trends and patterns in the education sector. The analysis helps in understanding changes in student admissions, institutional performance, gender distribution, and employment outcomes over the years. The insights generated from this project can support educational institutions, researchers, and policymakers in making informed decisions to improve the quality and effectiveness of higher education in India

```python
import pandas as pd
import numpy as np

# Re-loading the original dataset again to start fresh
url = "https://github.com/Aarthicodes/Indias-State-by-State-admission-and-placement-data-Analysis/raw/refs/heads/main/India_s%20state-by-state%20admission%20and%20placement%20data.xlsx"
df_refined = pd.read_excel(url, engine="openpyxl")

print("Original dataset reloaded for refined cleaning.")
```

```python
# Convert 'Year' column to datetime
df_refined['Year'] = pd.to_datetime(df_refined['Year'], format='%Y')
print("'Year' column converted to datetime.")
```

```python
# Identify numerical columns for outlier treatment
numerical_cols_for_outliers = df_refined.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Exclude 'Year' if it's still in numerical_cols_for_outliers (as it's now datetime, it likely won't be)
if 'Year' in numerical_cols_for_outliers:
    numerical_cols_for_outliers.remove('Year')

# Function to apply clipping within groups
def apply_clipping_by_group(group):
    for col in numerical_cols_for_outliers:
        if col in group.columns:
            Q1 = group[col].quantile(0.25)
            Q3 = group[col].quantile(0.75)
            IQR = Q3 - Q1

            lower = Q1 - 1.5 * IQR
            upper = Q3 + 1.5 * IQR

            group[col] = group[col].clip(lower=lower, upper=upper)
    return group

# Apply outlier treatment by 'State' group
df_refined = df_refined.groupby('State').apply(apply_clipping_by_group)

print("Outlier treatment applied by 'State' group for numerical columns.")
```

```python
# Re-create derived columns after refined outlier treatment
df_refined['Placement_Rate_Percentage'] = (df_refined['Placement'] / df_refined['Enrollment']) * 100
df_refined['Pass_ Rate_Percentage'] = (df_refined['Passed'] / df_refined['Enrollment']) * 100
df_refined['Student to Faculty Ratio'] = df_refined['Enrollment'] / df_refined['Faculties']
df_refined['Girls_Enrollment_Percentage'] = (df_refined['Girls_enrollment'] / df_refined['Enrollment']) * 100
df_refined['Boys_Enrollment_Percentage'] = (df_refined['Boys_enrolment'] / df_refined['Enrollment']) * 100

print("Derived columns re-created after refined cleaning.")
```

```python
# Display the head of the refined DataFrame to check the changes
print("Head of the refined DataFrame:")
display(df_refined.head())
```

```python
# Saving the refined DataFrame to a new CSV file
output_filename_refined = 'cleaned_higher_education_data_by_state.csv'
df_refined.to_csv(output_filename_refined, index=False)

print(f"Your refined cleaned dataset has been saved to '{output_filename_refined}'")
```
